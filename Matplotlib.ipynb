```python
#import matplotlib
import matplotlib.pyplot as plt
```

```python
#line plot.
x = [1,2,3,4,5]
y = [10,20,25,30,40]
```

```python
plt.plot(x,y)

plt.xlabel('This is X-Label')
plt.ylabel('This is Y-Label')
plt.title('This is our first line plot')

plt.show()
```

```python
plt.figure(figsize=(6,4))

x = [1,2,3,4,5]
y1 = [1,4,9,16,25]
y2 = [1,2,3,4,5]

#line plot
plt.plot(x,y1)
plt.plot(x,y2)

#line 1 plot
plt.plot(x,y1,label = "Line 1",color="pink")
plt.plot(x,y2,label = "Line 2",color="red")
plt.legend()
plt.grid()
```

```python
# Sample data
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
```

```python
fig = plt.figure(figsize=(6,4))
axes1 = fig.add_axes([0.1,0.1,0.8,0.8]) #[left,bottom,width,height]
axes1.plot(x,y)
plt.title('First plot')
axes2 = fig.add_axes([0.2,0.5,0.3,0.3])
axes2.plot(y,x)
plt.title('Second plot')
plt.show()
```

```python
#DPI

x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.figure(figsize=(6,4),dpi=100)  # dpi = controls zoom in and zoom out for a graph
plt.plot(x,y)
plt.grid()
```

```python
x = [1,2,3,4,5]
y1 = [2,4,6,8,10]
y2 = [1,3,5,7,9]

plt.plot(x,y1,label='Nili line',color='Black',lw=2.5)
plt.plot(x,y2,label='Orange line',lw=1.5,ls='-.',marker='*') #linewidth = lw , linestyle = ls

plt.title('Graph')
plt.legend()
```

```python
#Histogram plot -> shows spread of only one coloum/variables
import numpy as np
nums = np.random.randn(100)
plt.hist(nums,bins =20,)
```

```python
#Boxplot -> Shows outliers of a single column
plt.boxplot(nums)

#min
#25% -> lower Quatile
#50% -> Median
#75% -> Upper Quartile
#max

# values between lower and upper quatile are genric or actual values
# Values outside of lower and upper quartile are considered to be outliers
```
